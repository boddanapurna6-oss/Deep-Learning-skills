```python
import os
import sys
import torch
import matplotlib.pyplot as plt
import subprocess

print("Python Version:", sys.version)
print("PyTorch Version:", torch.__version__)
print("CUDA Available:", torch.cuda.is_available())
```

```python
project = "VisionPort"

folders = [
    "src",
    "models",
    "data",
    "notebooks",
    "outputs",
    "tests"
]

os.makedirs(project, exist_ok=True)

for folder in folders:
    os.makedirs(f"{project}/{folder}", exist_ok=True)

print("Project Repository Created!")

for root, dirs, files in os.walk(project):
    level = root.replace(project, "").count(os.sep)
    print("  " * level + os.path.basename(root) + "/")
```

```python
readme = """
# VisionPort

AI Model Development Environment

## Components
- Python
- PyTorch
- CUDA
- Git

## Project Structure
- src: Source code
- models: AI models
- data: Dataset
- notebooks: Experiments
- outputs: Results
- tests: Testing
"""

requirements = """
torch
torchvision
matplotlib
"""

with open(f"{project}/README.md", "w") as f:
    f.write(readme)

with open(f"{project}/requirements.txt", "w") as f:
    f.write(requirements)

print("README.md and requirements.txt created!")
```

```python
import torch

print("PyTorch Version:", torch.__version__)
print("CUDA Available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU Name:", torch.cuda.get_device_name(0))
    print("CUDA Version:", torch.version.cuda)

    device = torch.device("cuda")
else:
    print("Using CPU")
    device = torch.device("cpu")
```

```python
a = torch.randn(1000, 1000).to(device)
b = torch.randn(1000, 1000).to(device)

result = torch.matmul(a, b)

print("Device:", device)
print("Matrix Shape:", result.shape)
print("GPU Computation Successful!")
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

fig, ax = plt.subplots(figsize=(12, 4))
ax.set_xlim(0, 12)
ax.set_ylim(0, 4)
ax.axis("off")

steps = [
    ("git init", "Initialize"),
    ("git add", "Stage Files"),
    ("git commit", "Save Changes"),
    ("git push", "Upload to GitHub")
]

colors = ["#4A90E2", "#9067C6", "#35A879", "#E58B45"]
positions = [0.3, 3.2, 6.1, 9]

for i, (title, desc) in enumerate(steps):
    box = FancyBboxPatch(
        (positions[i], 1.3), 2.3, 1.4,
        boxstyle="round,pad=0.15",
        facecolor=colors[i],
        edgecolor="black"
    )
    ax.add_patch(box)

    ax.text(positions[i]+1.15, 2.15, title,
            ha="center", color="white", weight="bold")
    ax.text(positions[i]+1.15, 1.75, desc,
            ha="center", color="white")

    if i < 3:
        ax.annotate("", xy=(positions[i+1]-0.15, 2),
                    xytext=(positions[i]+2.4, 2),
                    arrowprops=dict(arrowstyle="->", lw=2))

plt.title("Git Development Workflow", fontsize=16, weight="bold")
plt.savefig("git_workflow.png", dpi=200)
plt.show()
```

```python
components = ["Python", "PyTorch", "CUDA", "GPU"]
status = [
    1,
    1,
    int(torch.version.cuda is not None),
    int(torch.cuda.is_available())
]

plt.figure(figsize=(8, 4))
plt.barh(components, status)
plt.xlim(0, 1.2)
plt.xticks([0, 1], ["Not Available", "Available"])
plt.title("PyTorch–CUDA Environment Status")
plt.xlabel("Availability")
plt.grid(axis="x", alpha=0.3)
plt.show()
```
